```python
stack = []
while True:
    print(f"\nCurrent Stack:{stack}")
    print("1:push| 2.pop| 3.peek|4.Exit")
    choice = input("Enter choice(1-4):")

    if choice =="1":
        item=input("enter value to push:")
        stack.append(item)
    elif choice =="2":
        if stack:
            print(f"popped: {stack.pop()}")
        else:
            print("stack underflow empty stack")
    elif choice =="3":
        if stack:
            print(f"top element:{stack[-1]}")
        else:
            print("stack is empty")
    elif choice =="4":
        print("Goodbyee!!")
        break
    else:
        print("Invalid choice")
```
